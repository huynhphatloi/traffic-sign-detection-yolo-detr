# 04 · YOLO Baseline (plan §8.5)

Train YOLOv8n (pretrained, imgsz 640, flip-LR OFF) and evaluate. **Needs a GPU.** Owner: **Loi**. Writes `weights/yolo/yolo_baseline/best.pt` and `results/metrics/yolo_baseline.json`.

```python
# ── Session init (run this first on every new Colab machine) ────────────────
import os, sys, subprocess
from google.colab import drive

drive.mount('/content/drive')

GITHUB_REPO = 'https://github.com/huynhphtloi/traffic-sign-detection-yolo-detr'
DRIVE_ROOT  = '/content/drive/MyDrive/HCMUE_Projects/DeepLearning'
REPO        = '/content/traffic-sign-detection-yolo-detr'
DATA_DIR    = f'{DRIVE_ROOT}/data'

if not os.path.exists(REPO):
    subprocess.run(['git', 'clone', GITHUB_REPO, REPO], check=True)
else:
    subprocess.run(['git', '-C', REPO, 'pull'], check=True)

os.environ.update({'TSD_DRIVE_ROOT': DRIVE_ROOT, 'TSD_REPO_ROOT': REPO, 'TSD_DATA_DIR': DATA_DIR})
if REPO not in sys.path:
    sys.path.insert(0, REPO)

os.system(f'pip install -q -r {REPO}/requirements.txt')
print('ready — repo:', REPO, '| data:', DATA_DIR)
```

## Train

```python
from src.training.train_yolo import train
results, best = train(model='yolov8n.pt', epochs=30, name='yolo_baseline')
print('best checkpoint:', best)
```

## Evaluate on the test split (mAP, P, R, FPS, size) → metrics JSON

```python
from src.evaluation.evaluate_yolo import evaluate
from src.utils.paths import RESULTS_METRICS, ensure_dir
import json
metrics = evaluate(best, split='test')
ensure_dir(RESULTS_METRICS)
(RESULTS_METRICS / 'yolo_baseline.json').write_text(json.dumps(metrics, indent=2))
metrics
```
